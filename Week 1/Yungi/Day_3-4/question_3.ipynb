```python
"""
The prime factors of 13195 are 5,7,13, and 29.

What is the largest prime factor of the number 600851475143?

"""
```

```python
factor_of = 600851475143


def finding_factor(n):
    lst2 = []
    for i in range(1,n+1):
        if n % i == 0:
            lst2.append(i)
        else:
            continue
    return lst2

#[1,3,5,....]

def prime(a):
    prime_lst = []
    for i in range(len(a)):
        if len(finding_factor(a[i])) == 2:
            prime_lst.append(a[i])
        else:
            continue
    return prime_lst



def max_prime(b):
    answer = sorted(b)
    real_answer = answer[-1]
    return real_answer


ang = finding_factor(factor_of)
p = prime(ang)

answer = max_prime(p)
print(answer)
```
